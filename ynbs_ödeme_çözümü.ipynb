```python
rub_islemler = dataframe[dataframe['Döviz Kodu'] == 'RUB'].groupby('Issuer Ülke Kodu')['İşlem Tutarı (Döviz kodu cinsinden)'].sum()
jpy_islemler = dataframe[dataframe['Döviz Kodu'] == 'JPY'].groupby('Issuer Ülke Kodu')['İşlem Tutarı (Döviz kodu cinsinden)'].sum()

kur_tablosu = {
    'RUB': 0.011,  # 1 RUB = 0.011 euro (tahmini)
    'JPY': 0.0061   # 1 JPY =  0.0061 euro(tahmini)
}

rub_eu = rub_islemler * kur_tablosu['RUB']
jpy_eu = jpy_islemler * kur_tablosu['JPY']

print("RUB ile işlem yapan ülkelerin Euro cinsinden harcamaları:\n", rub_eu)
print("JPY ile işlem yapan ülkelerin Euro cinsinden harcamaları:\n", jpy_eu)
```

```python
import pandas as pd

# 1. Önce sütun isimlerini kontrol edelim
print("Mevcut sütunlar:", dataframe.columns.tolist())

# 2. Doğru sütun adını belirleyelim (örneğin 'İşlem Tutarı' olabilir)
islem_tutari_sutunu = "İşlem Tutarı (Döviz kodu cinsinden)"  # Verinizdeki gerçek sütun adıyla değiştirin

# 3. Döviz kurlarını tanımlayalım
doviz_kurlari = {
    "RUB": 0.011,  # 1 RUB = 0.011 USD
    "JPY": 0.0072, # 1 JPY = 0.0072 USD
    "USD": 1.0
}

# 4. RUB işlemlerini filtreleyip USD'ye çevirelim
rub_islemler = dataframe[dataframe["Döviz Kodu"] == "RUB"].copy()  # .copy() ekledik
rub_islemler["İşlem Tutarı (USD)"] = rub_islemler[islem_tutari_sutunu] * doviz_kurlari["RUB"]

# 5. JPY işlemlerini filtreleyip USD'ye çevirelim
jpy_islemler = dataframe[dataframe["Döviz Kodu"] == "JPY"].copy()  # Döviz Kodu mu Döviz Kodu mu kontrol edin
jpy_islemler["İşlem Tutarı (USD)"] = jpy_islemler[islem_tutari_sutunu] * doviz_kurlari["JPY"]

# 6. Gruplama ve sonuçları görüntüleme
rub_usd_toplam = rub_islemler.groupby("Issuer Ülke Kodu")["İşlem Tutarı (USD)"].sum().sort_values(ascending=False)
jpy_usd_toplam = jpy_islemler.groupby("Issuer Ülke Kodu")["İşlem Tutarı (USD)"].sum().sort_values(ascending=False)

print("RUB İşlemleri (USD):\n", rub_usd_toplam)
print("\nJPY İşlemleri (USD):\n", jpy_usd_toplam)
```

```python
import matplotlib.pyplot as plt

# RUB ve JPY hacimlerini karşılaştırmalı bar grafiği
plt.figure(figsize=(10, 6))
rub_usd_toplam.head(5).plot(kind='bar', color='red', alpha=0.5, label='RUB (USD)')
jpy_usd_toplam.head(5).plot(kind='bar', color='blue', alpha=0.5, label='JPY (USD)')
plt.title("Ülkelere Göre RUB ve JPY İşlem Hacimleri (USD)")
plt.ylabel("Toplam İşlem Tutarı (USD)")
plt.legend()
plt.show()
```
